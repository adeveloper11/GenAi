### Task-3 (product pricing - Dictioniries)

```python
# 1)
price_dict = {
    "phone": 12566,
    "books": 55000,
    "clothes": 5000,
    "food": 2000,
    "toys": 4000,
    "bag": 1200
}
price_dict
```

```python
# 2)
price_dict["charger"] = 1100 #added new object
price_dict["toys"] = 4500 #changed Price
try:
    price_dict.pop("bag") # remove object
except Exception as e:
    print(f"object not found error : {e}")
```

```python
price_dict
```

```python
# 3
Total_sum = 0
count = 0
for k,v in price_dict.items():
    count += 1
    Total_sum += v

average_price = Total_sum/count
average_price
```

```python
# Extra
price = []
for k,v in price_dict.items():
    price.append(v)
print(min(price))
print(max(price))
        
```

### Task -4 Combined Operations

```python
# 1) Create a catalog list of tuples
catalog = []
for product in Products:
    product_key = product.lower()
    if product_key in price_dict:
        category = "electronics" if product_key in ["phone", "laptop", "speaker", "watch", "charger"] else "other"
        catalog.append((product, price_dict[product_key], category))

catalog
```

```python
# 2
category_to_products = {}
for product_name, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []
    category_to_products[category].append((product_name, price))

category_to_products
```

```python
# 3
for keys,values in category_to_products.items():
    print(len(values))
```

```python
# 4
max_category = max(category_to_products, key=lambda category: len(category_to_products[category]))
print(f"Category with maximum products: {max_category}")
for product_name, price in category_to_products[max_category]:
    print(product_name)
```
