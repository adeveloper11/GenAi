```python
## recursive Function
def factorialfun(num):
    if num == 0 or num == 1:
        print(1)
    else:
        factorial = 1
        for i in range(1, num + 1): # multiply digit to  start from 1 to upto given number.
            factorial = factorial*i
        print(factorial)

factorialfun(13)
```

```python
## Recursive function
def recursive_factorial(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * recursive_factorial(num -1)
```

```python
recursive_factorial(3)
```
