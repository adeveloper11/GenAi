#### Task-3 user menu (while loop + break/continue)

```python
# 1
order_amount = []
total_order_amount = 0
print("MENU :")
print("1 for add order amount :")
print("2 show all orders and discount applied:")
print("q to quit")

while True:
    user_input = input("ENTER VALUE FROM GIVEN MENU : ")
    # 1.1 adding to order amount
    if user_input == "1":
        amount = input("Enter Amount")
        order_amount.append(amount)
    # 1.2 show all orders and discount after applying discount
    elif user_input == "2":
        total = sum(int(order) for order in order_amount)
        # Calculate discount inline
        if total >= 2000:
            discount_percentage = 15
        elif total >= 1500:
            discount_percentage = 10
        elif total >= 1000:
            discount_percentage = 7
        else:
            discount_percentage = 0
        discount_amount = discount_percentage/100 * total
        final_amount = total - discount_amount
        print(f"all orders : {order_amount}")
        print(f"total amount : {total}")
        print(f"discount percentage applied : {discount_percentage}%")
        print(f"discount amount : {discount_amount}")
        print(f"final amount after discount : {final_amount}")
    # quit
    elif user_input == "q":
        break
    else:
        print("please Enter valid data")
    
```
