### Task-5 Loop control with condition

```python
# 1
daily = [200, 150, 0, 400, 50, -1, 300]
```

```python
#1.
total_sales = 0
corrupted_data_count = 0
for amount in daily:
    if amount == -1:
        #treat it as corrupted data and break loop
        corrupted_data_count +=1
        break
    elif amount == 0:
        #skips revenue
        continue
    else:
        #valid sales added to total sales
        total_sales += amount
        print(f"Sale: {amount}, Running Total: {total_sales}")
        
        
    
```

```python
#2
print("Final amount after loop complete or stops due to corrupted data is :",total_sales)
```
