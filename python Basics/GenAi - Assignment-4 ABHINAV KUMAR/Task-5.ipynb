### Task -5 Create product into file

```python
products = []
```

```python
#1. Ask the user for 3 product names & their prices.
for i in range (1,4):
    name = str(input(f"Enter product {i} name"))
    price = float(input(f"Enter product {i} price"))
    products.append((name, price))
```

```python
#2. Write them into a new file products.txt in this format:
with open ("products.txt", "w") as file:
    for index, (name, price) in enumerate(products, start=1):
        file.write(f"{index}. {name} | {price}\n")
    
```

```python
#3. ProductName | Price
with open ("products.txt", "w") as file:
    for index, (name, price) in enumerate(products, start=1):
        file.write(f"{index}. {name} | {price}\n")
```

```python
# Read the file and print each line with propper formeting
with open("products.txt", "r") as file:
    for line in file:
        line = line.strip()
        idx_id, price_str = line.split("|")
        idx_id, name = idx_id.split(".")
        price = float(price_str)
        print(f"{idx_id} {name} {price}")       
        
```
