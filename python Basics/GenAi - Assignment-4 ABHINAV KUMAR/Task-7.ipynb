### Task-7 mini project Export Discounted Prices

```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

discount_percent = float(input("Enter discount percentage: "))

header = "Product | Original Price | Discounted Price"
lines = [header]

for product, price in prices.items():
    discounted_amount = (price*discount_percent)/100
    discounted_price = price - discounted_amount
    lines.append(f"{product} | {price} | {discounted_price}")

with open("discount_report.txt", "w") as file:
    file.write("\n".join(lines) + "\n")

with open("discount_report.txt", "r") as file:
    print(file.read())
```
