```python
#4 write each sale on a new line in the file
f = open("sales_data.txt", "w")
for sale in sales:
    f.write(str(sale)+'\n')
f.close()
```

```python
# 5 close the file and reopen it and print its contents
f.close()
f = open("sales_data.txt", "r")
file_data = f.read()
print(file_data)
f.close()
```
