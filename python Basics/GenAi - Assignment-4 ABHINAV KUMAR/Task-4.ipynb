```python
#3.1
total_sales = 0
with open("sales_data.txt", "r") as f:
    for line in f:
        for part in line.strip().split(","):
            if part:
                value = int(part)
                total_sales += value
print("Total Sales = ",total_sales)
```

```python
#3.2 Highest sales
sales_data = []
with open("sales_data.txt", "r") as f:
    for line in f:
        for part in line.strip().split(","):
            if part:
                value = int(part)
                sales_data.append(value)
Highest_sales = max(sales_data)
print("Highest_sales:", Highest_sales)
```

```python
#3.3
lowest_sales = min(sales_data)
print(lowest_sales)
```

```python
#3.4

average_sales = sum(sale for sale in sales_data)/len(sales_data)
print(average_sales)
```
