### Task-2 Recursive Function Factorital utility

```python
# Helper function
def factorial(n):
    if n < 0:
        return print("Enter valid number for factorial")
    if n <= 1:
        return 1
    return n * factorial(n-1)
    
```

```python
# 1
factorial(5)
```

```python
# 2
factorial(0)
```

```python
factorial(1)
```
