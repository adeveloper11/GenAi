### Task-7 Mini problem : Menu using function

```python
#Helper Function
price_list = []
def add_price(price = 0):
    price_list.append(price)
    return price_list
def get_average_price(price_list):
    length_of_price_list = len(price_list)
    total_price = 0
    for price in price_list:
        total_price += price
        average_price = total_price/length_of_price_list
    return average_price
def get_max_price(price_list):
    maximum = max(price_list)
    return maximum
```

```python

while True:
    print("Enter 1 to Add Price")
    print("Enter 2 to Show Average price")
    print("Enter 3 to shoe Highest price")
    print("Enter q to Quit")
    user_input = input("ENTER OPTION FROM MENU : ")
    if user_input == "q":
        price_list = []
        break
#1
    if user_input == "1":
        added_price = float(input("Enter price"))
        user_added_price = add_price(added_price)
        print(user_added_price)
#2
    elif user_input == "2":
        if price_list:
            average_price = get_average_price(price_list)
            print(average_price)
#3
    elif user_input == "3":
        maximum_price = get_max_price(price_list)
        print(maximum_price)
    else:
        print("Enter Option from MENU :")
        continue
    
        
```
