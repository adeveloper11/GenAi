### using Map() : Apply discount on list of prices

```python
original_prices = [100, 250, 400, 1200, 50]
```

```python
prices_with_gst = list(map(lambda price:price + (.18* price), original_prices))
print(prices_with_gst)
```

```python
print("original_prices : ",original_prices)
print("prices_with_gst : ",prices_with_gst)
```
