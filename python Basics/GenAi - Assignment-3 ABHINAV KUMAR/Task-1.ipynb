### Task - 1 - Basic Function Price after discount

```python
# Heler function
def apply_discount(price=0, discount_percent=5):
        if discount_percent > 60:
            discount_percent = 60
        discount_applied = discount_percent/100 * price
        discounted_amount = price - discount_applied
        return discounted_amount
```

```python
#1
price_after_discount = apply_discount(200,15)
print(price_after_discount)
```

```python
#2
missing_discount_percent = apply_discount(500)
print(missing_discount_percent)
```
