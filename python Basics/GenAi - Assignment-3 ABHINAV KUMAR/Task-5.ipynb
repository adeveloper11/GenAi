### Task-5 Using Filter() Filter expensive products

```python
original_prices = [100, 250, 400, 1200, 50, 2000, 850]
```

```python
# 1
greater_then_500 = list(filter(lambda price:price > 500, original_prices))
```

```python
# 2
less_then_or_equal_to_500 = list(filter(lambda price:price <= 500, original_prices))
```

```python
print("original_prices : ", original_prices)
print("greater_then_500 :", greater_then_500)
print("less_then_or_equal_to_500 :", less_then_or_equal_to_500)
```
